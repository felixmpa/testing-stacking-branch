**SVM - Fit regression model** :: SVR

```python
# Authors: The scikit-learn developers/adaptado por Sdelorbe 2025
# SPDX-License-Identifier: BSD-3-Clause

import matplotlib.pyplot as plt
import numpy as np

from sklearn.svm import SVR
from sklearn.svm import SVR
from sklearn.metrics import  mean_squared_error, roc_auc_score, accuracy_score, r2_score

# random data
X = np.sort(5 * np.random.rand(40, 1), axis=0)
y = np.sin(X).ravel()

display(X)
display(y)

# add noise to targets/ruidos
y[::5] += 3 * (0.5 - np.random.rand(8))
svr_rbf = SVR(kernel="rbf", C=100, gamma=0.1, epsilon=0.1)
svr_lin = SVR(kernel="linear", C=100, gamma="auto")
svr_poly = SVR(kernel="poly", C=100, gamma="auto", degree=3, epsilon=0.1, coef0=1)
# " Look at the results
lw = 2

svrs = [svr_rbf, svr_lin, svr_poly]
kernel_label = ["RBF", "Linear", "Polynomial"]
model_color = ["m", "c", "g"]

fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(15, 10), sharey=True)
for ix, svr in enumerate(svrs):
    yp = svr.fit(X, y).predict(X)
    axes[ix].plot(
        X,
        yp,
        color=model_color[ix],
        lw=lw,
        label="{} model".format(kernel_label[ix]),
    )
    axes[ix].scatter(
        X[svr.support_],
        y[svr.support_],
        facecolor="none",
        edgecolor=model_color[ix],
        s=50,
        label="{} support vectors".format(kernel_label[ix]),
    )
    axes[ix].scatter(
        X[np.setdiff1d(np.arange(len(X)), svr.support_)],
        y[np.setdiff1d(np.arange(len(X)), svr.support_)],
        facecolor="none",
        edgecolor="k",
        s=50,
        label="other training data",
    )
    axes[ix].legend(
        loc="upper center",
        bbox_to_anchor=(0.5, 1.1),
        ncol=1,
        fancybox=True,
        shadow=True,
    )
    print(f"Error cuadrático medio (MSE) de SVR-{kernel_label[ix]}: {mean_squared_error(y, yp):.2f}")
    print(f"R2 % SVR-{kernel_label[ix]}: {r2_score(y, yp)*100:.2f}")

fig.text(0.5, 0.04, "data", ha="center", va="center")
fig.text(0.06, 0.5, "target", ha="center", va="center", rotation="vertical")
fig.suptitle("Support Vector Regression", fontsize=14)
plt.show()
```

SVM por clasificacion/ dataset gnerado

```python
import matplotlib.pyplot as plt

from sklearn import svm
from sklearn.datasets import make_blobs
from sklearn.inspection import DecisionBoundaryDisplay

# we create 40 separable points
X, y = make_blobs(n_samples=40, centers=2, random_state=6)

display(X)
display(y)


# fit the model, don't regularize for illustration purposes
clf = svm.SVC(kernel="linear", C=1000)
clf.fit(X, y)

plt.scatter(X[:, 0], X[:, 1], c=y, s=30, cmap=plt.cm.Paired)

# plot the decision function
ax = plt.gca()
DecisionBoundaryDisplay.from_estimator(
    clf,
    X,
    plot_method="contour",
    colors="k",
    levels=[-1, 0, 1],
    alpha=0.5,
    linestyles=["--", "-", "--"],
    ax=ax,
)
# plot support vectors
ax.scatter(
    clf.support_vectors_[:, 0],
    clf.support_vectors_[:, 1],
    s=100,
    linewidth=1,
    facecolors="none",
    edgecolors="k",
)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
import seaborn as sns

# Load data
iris = load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)

display(X.head(10))

# Standardize
scaler = StandardScaler()
X_std = scaler.fit_transform(X)
### display(X_std)

# PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_std)
print("Componentes principales:\n", pca.components_)

# Scree plot (Explained Variance)
explained_variance = pca.explained_variance_ratio_
cum_var = np.cumsum(explained_variance)

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(cum_var)+1), cum_var, marker='o')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Scree Plot')
plt.grid(True)
plt.show()
# Crear un DataFrame para las cargas (loadings)
loadings_df = pd.DataFrame(
    pca.components_.T,
    columns=[f'PC{i+1}' for i in range(pca.n_components_)],
    index=X.columns
)

print("Matriz de cargas:")
print(loadings_df)
# Identificar las variables más influyentes para cada componente
for pc in loadings_df.columns:
    most_influential_variable = loadings_df[pc].abs().idxmax()
    influence_value = loadings_df.loc[most_influential_variable, pc]
    print(f"\nLa variable más influyente en {pc} es '{most_influential_variable}' con un peso de {influence_value:.2f}")



# También puedes ver la varianza explicada por cada componente para entender su importancia
# explained_variance = pca.explained_variance_ratio_
print("\nVarianza explicada por cada componente:")
print(explained_variance)

# PROCESO DE VISUALIZACION:: Crear un DataFrame con los dos componentes principales
pc_df = pd.DataFrame(data=X_pca, columns=['PC1', 'PC2'])

# Obtener los loadings (cargas) de las variables
loadings = pca.components_.T * np.sqrt(pca.explained_variance_)
loadings_df = pd.DataFrame(
    loadings,
    columns=['PC1', 'PC2'],
    index=X.columns
)

# Configurar el gráfico
plt.figure(figsize=(10, 8))
sns.scatterplot(x='PC1', y='PC2', data=pc_df, s=100)
plt.title('Biplot de PCA')
plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.2f}% de varianza explicada)')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.2f}% de varianza explicada)')

# Agregar los vectores (flechas) para las variables
for i, variable in enumerate(loadings_df.index):
    plt.arrow(
        0, 0,
        loadings_df.loc[variable, 'PC1'],
        loadings_df.loc[variable, 'PC2'],
        head_width=0.05,
        head_length=0.05,
        color='r',
        alpha=0.8
    )
    plt.text(
        loadings_df.loc[variable, 'PC1']*1.15,
        loadings_df.loc[variable, 'PC2']*1.15,
        variable,
        color='b',
        ha='center',
        va='center',
        fontsize=12
    )

plt.grid()
plt.axhline(0, color='gray', linestyle='--')
plt.axvline(0, color='gray', linestyle='--')
plt.show()
```

```python
import pandas as pd
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler, LabelEncoder
import matplotlib.pyplot as plt
import seaborn as sns
import zipfile
import requests
from io import BytesIO
import os

# --- 1. Descarga y Carga de Datos ---
# URL del archivo zip del dataset OULAD
url = "https://raw.githubusercontent.com/aishwaryar7/Open-University-Learning-Analysis/master/anonymisedData/studentInfo.csv"
print("Descargando el dataset OULAD...")

# Descargar el contenido del archivo zip
# r = requests.get(url)
# z = zipfile.ZipFile(BytesIO(r.content))

# Extraer y cargar los archivos necesarios en dataframes de pandas
print("Cargando los datos...")
#with z.open('studentInfo.csv') as f:
student_info = pd.read_csv(url)

# --- 2. Preparación de los Datos ---
# Seleccionamos las características que usaremos para el t-SNE
# Usaremos una combinación de variables categóricas y numéricas
features = student_info[[
    'gender',
    'region',
    'highest_education',
    'imd_band',
    'age_band',
    'num_of_prev_attempts',
    'studied_credits',
    'disability',
    'final_result' # Usaremos esto como etiqueta para colorear, no como característica
]].copy()

# Eliminar filas donde el resultado final es 'Withdrawn' para una visualización más clara
features = features[features['final_result'] != 'Withdrawn']

# Guardar las etiquetas (resultado final) para la visualización
labels = features['final_result']
features_to_encode = features.drop('final_result', axis=1)

# --- 3. Preprocesamiento ---
# Convertir variables categóricas a numéricas usando one-hot encoding
print("Preprocesando los datos...")
features_encoded = pd.get_dummies(features_to_encode, drop_first=True)

# Escalar las características numéricas para que tengan media 0 y desviación estándar 1
# Esto es importante para que t-SNE funcione correctamente
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features_encoded)

# --- 4. Aplicación de t-SNE ---
# Inicializamos t-SNE. Los parámetros más comunes a ajustar son:
# - n_components: La dimensionalidad del espacio de salida (casi siempre 2 para visualizar).
# - perplexity: Relacionado con el número de vecinos cercanos. Un valor entre 5 y 50 es típico.
# - n_iter: Número de iteraciones para la optimización.
print("Aplicando t-SNE... (esto puede tardar unos minutos)")
tsne = TSNE(n_components=2, perplexity=30, n_iter=1000, random_state=42)
tsne_results = tsne.fit_transform(features_scaled)

# Crear un DataFrame con los resultados de t-SNE para facilitar la visualización
tsne_df = pd.DataFrame(
    data=tsne_results,
    columns=['tsne_component_1', 'tsne_component_2']
)
tsne_df['final_result'] = labels.values

# --- 5. Visualización ---
print("Generando la visualización...")
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(12, 8))

# Crear el gráfico de dispersión usando seaborn
scatter = sns.scatterplot(
    x='tsne_component_1',
    y='tsne_component_2',
    hue='final_result',
    palette=sns.color_palette("hls", len(tsne_df['final_result'].unique())),
    data=tsne_df,
    legend="full",
    alpha=0.7,
    ax=ax
)

ax.set_title('Visualización t-SNE del Dataset OULAD por Resultado Final', fontsize=16)
ax.set_xlabel('Componente t-SNE 1', fontsize=12)
ax.set_ylabel('Componente t-SNE 2', fontsize=12)
ax.legend(title='Resultado Final')

plt.show()
```
